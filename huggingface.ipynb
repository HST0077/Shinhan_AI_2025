# **감성분석**

'I love to communicate with Koreans.' 문장이 긍정인지 부정인지 판단하여라.

```python
from transformers import pipeline
sentiment=pipeline('sentiment-analysis')
```

```python
sentiment.model
```

```python
sentiment('I love to communicate with Koreans')
```

나는 이 제품이 정말 좋아요.’ 라는 문장이 긍정인지 부정인지 판단하여라.

```python
sentiment('나는 이 제품이 정말 좋아요')
```

```python
from transformers import pipeline

kor_sentiment=pipeline('text-classification',
                       model='tabularisai/multilingual-sentiment-analysis')
kor_sentiment('나는 이 제품이 정말 좋아요.')
```

```python
kor_sent02=pipeline(
    task='text-classification',
    model='WhitePeak/bert-base-cased-korean-sentiment',
)
kor_sent02('나는 이 제품이 정말 좋아요.')
```
