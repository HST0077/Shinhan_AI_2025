```python
import re
```

```python
T0='1. 이메일 주소는 user@example.com과 admin@example.org입니다. \n2. 전화번호는 123-4567-0000 입니다'
print(T0)
```

```python
# T0에서 숫자를 모두 삭제 
T1=re.sub(r'\d','',T0)
print(T1)
```

```python
# T0에서 1,2,3만 삭제
T2=re.sub(r'[1-3]','',T0)
print(T2)
```

```python
# T0에서 a,b,c만 삭제
T3=re.sub(r'[a-c]','',T0)
print(T3)
```

```python
T='이메일 주소는 user@example.com과 admin@example.org입니다.'
pattern='@'
# 문자열에서 패턴을 추출해서 리스트로 반환
re.findall(pattern,T) 
```

```python
# 문자열에서 패턴을 모두 찾아 iterateor로 반환
D=re.finditer(pattern, T)
for item in D:
    print(item)
```

```python
# 첫 번째 부분 문자열의 위치를 반환
re.search(pattern, T)
```

```python
re.match(pattern, T)
```
